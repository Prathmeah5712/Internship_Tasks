## Q1. Find the Shortest Word in a Sentence

```python
def find_shortest_word(sentence):
    words = sentence.split()
    if not words:
        return ""
    shortest = words[0]
    for w in words[1:]:
        if len(w) < len(shortest):
            shortest = w
    return shortest

print(find_shortest_word("Python makes coding easy"))
print(find_shortest_word("Data science is a broad domain"))
```

## Q2. Find the Largest Number in a List (without max())

```python
def find_largest(numbers):
    if not numbers:
        return None
    largest = numbers[0]
    for num in numbers[1:]:
        if num > largest:
            largest = num
    return largest

print(find_largest([12, 45, 7, 89, 23]))
print(find_largest([-10, -5, -20, -1]))
```

## Q3. Find the Smallest Number in a List (without min())

```python
def find_smallest(numbers):
    if not numbers:
        return None
    smallest = numbers[0]
    for num in numbers[1:]:
        if num < smallest:
            smallest = num
    return smallest

print(find_smallest([12, 45, 7, 89, 23]))
print(find_smallest([100, 54, 2, 88, 15]))
```

## Q6. Count Vowels in a String

```python
def count_vowels(text):
    vowels = set("aeiouAEIOU")
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

print(count_vowels("Education"))
print(count_vowels("Natural Language Processing"))
```

## Q7. Count Consonants in a String

```python
def count_consonants(text):
    vowels = set("aeiouAEIOU")
    count = 0
    for ch in text:
        if ch.isalpha() and ch not in vowels:
            count += 1
    return count

print(count_consonants("Python 123!"))
print(count_consonants("Hello World"))
```

## Q8. Find the Most Frequent Character

```python
def most_frequent_char(text):
    freq = {}
    for ch in text:
        if ch != ' ':
            freq[ch] = freq.get(ch, 0) + 1
    if not freq:
        return None
    most_char = None
    max_count = 0
    for ch in freq:
        if freq[ch] > max_count:
            max_count = freq[ch]
            most_char = ch
    return most_char

print(most_frequent_char("banana"))
print(most_frequent_char("data analytics"))
```

## Q9. Find the Least Frequent Character

```python
def least_frequent_char(text):
    freq = {}
    for ch in text:
        if ch != ' ':
            freq[ch] = freq.get(ch, 0) + 1
    if not freq:
        return None
    least_char = None
    min_count = float('inf')
    for ch in freq:
        if freq[ch] < min_count:
            min_count = freq[ch]
            least_char = ch
    return least_char

print(least_frequent_char("success"))
print(least_frequent_char("engineering"))
```

## Q10. Find All Characters Appearing More Than Once

```python
def repeated_characters(text):
    freq = {}
    for ch in text:
        if ch != ' ':
            freq[ch] = freq.get(ch, 0) + 1
    return [ch for ch, count in freq.items() if count > 1]

print(repeated_characters("programming"))
print(repeated_characters("statistics"))
```

## Q15. Dictionary of Numbers and Their Even/Odd Status

```python
def even_odd_dict(numbers):
    result = {}
    for num in numbers:
        if num % 2 == 0:
            result[num] = "Even"
        else:
            result[num] = "Odd"
    return result

print(even_odd_dict([1, 2, 3, 4, 5]))
print(even_odd_dict([10, 15, 22, 33]))
```

## Q16. Find the Most Frequent Word in a Sentence

```python
def most_frequent_word(sentence):
    words = [w.lower() for w in sentence.split()]
    if not words:
        return ""
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    most_word = None
    max_count = 0
    for w in freq:
        if freq[w] > max_count:
            max_count = freq[w]
            most_word = w
    return most_word

print(most_frequent_word("Python is easy and Python is powerful"))
print(most_frequent_word("Data is gold and data is power"))
```

## Q17. Find All Words Appearing More Than Once

## Q18. Find the First Word That Appears Only Once

```python
def first_unique_word(sentence):
    words = [w.lower() for w in sentence.split()]
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    for w in words:
        if freq[w] == 1:
            return w
    return None

print(first_unique_word("apple banana apple mango banana orange"))
print(first_unique_word("tea coffee tea milk coffee"))
```

## Q20. Find the Longest Word That Appears Only Once

```python
def longest_unique_word(sentence):
    words = [w.lower() for w in sentence.split()]
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    unique_words = []
    seen = set()
    for w in words:
        if freq[w] == 1 and w not in seen:
            unique_words.append(w)
            seen.add(w)
    if not unique_words:
        return None
    longest = unique_words[0]
    for w in unique_words[1:]:
        if len(w) > len(longest):
            longest = w
    return longest

print(longest_unique_word("Python makes coding useful Python"))
print(longest_unique_word("data data analysis pipeline models"))
```
