#### Q1. Find the Frequency of Each Word in a Sentence
Pseudocode:

Initialize an empty dictionary counts.

Split the sentence into individual words using space as delimiter.

Iterate through each word:

If word exists in counts, increment its value by 1.

Else, add word with value 1.

Return counts.

```python
def word_frequency(sentence):
    counts = {}
    for word in sentence.split():
        counts[word] = counts.get(word, 0) + 1
    return counts

# Given Test Case
print("Given Input Output:     ", word_frequency("python is easy and python is powerful"))

# Additional Test Case
print("Additional Input Output:", word_frequency("data science is fun and python is fun"))
```

#### Q2. Find the Frequency of Each Character in a Word
Pseudocode:

Initialize an empty dictionary counts.

Iterate through each character in the given word:

If character exists in counts, increment its value by 1.

Else, add character with value 1.

Return counts.

```python
def char_frequency(word):
    counts = {}
    for ch in word:
        counts[ch] = counts.get(ch, 0) + 1
    return counts

# Given Test Case
print("Given Input Output:     ", char_frequency("banana"))

# Additional Test Case
print("Additional Input Output:", char_frequency("internship"))
```

#### Q3. Find the First Non-Repeating Character in a Word
Pseudocode:

Initialize an empty dictionary counts.

Traverse through each character in the word and store its occurrence count.

Traverse the word again from left to right:

If the count of the character is 1, return this character immediately.

If no such character is found, return None.

```python
def first_non_repeating_char(word):
    counts = {}
    for ch in word:
        counts[ch] = counts.get(ch, 0) + 1
    for ch in word:
        if counts[ch] == 1:
            return ch
    return None

# Given Test Case
print("Given Input Output:     ", first_non_repeating_char("swiss"))

# Additional Test Case
print("Additional Input Output:", first_non_repeating_char("aabbcc"))
```

#### Q4. Check Whether Two Strings Are Anagrams
Pseudocode:

Sort the characters of both strings s1 and s2.

Compare the sorted versions of both strings.

If both sorted strings are identical, return True.

Otherwise, return False.

```python
def are_anagrams(s1, s2):
    return sorted(s1) == sorted(s2)

# Given Test Case
print("Given Input Output:     ", are_anagrams("listen", "silent"))

# Additional Test Case
print("Additional Input Output:", are_anagrams("triangle", "integral"))
```

#### Q5. Find the First Repeating Character in a String
Pseudocode:

Initialize an empty set seen to store visited characters.

Iterate through each character in the string from left to right:

If the character is already present in seen, return this character.

Else, add the character to seen.

If the loop completes without finding any repeating character, return None.

```python
def first_repeating_char(s):
    seen = set()
    for ch in s:
        if ch in seen:
            return ch
        seen.add(ch)
    return None

# Given Test Case
print("Given Input Output:     ", first_repeating_char("programming"))

# Additional Test Case
print("Additional Input Output:", first_repeating_char("abcdef"))
```

#### Q6. Move All Zeros to the End of a List
Pseudocode:

Filter out all non-zero elements from the list while keeping their relative order.

Count the number of zeros in the original list.

Create a list containing the counted number of zeros.

Concatenate the non-zero list with the zero list and return it.

```python
def move_zeros_to_end(lst):
    non_zeros = [x for x in lst if x != 0]
    zeros = [0] * (len(lst) - len(non_zeros))
    return non_zeros + zeros

# Given Test Case
print("Given Input Output:     ", move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))

# Additional Test Case
print("Additional Input Output:", move_zeros_to_end([1, 0, 2, 0, 3, 0]))
```

#### Q7. Find the Second Highest Distinct Number in a List
Pseudocode:

Convert the list to a set to remove all duplicate numbers.

Convert the set back to a list and sort it in descending order.

Check the length of the sorted distinct list:

If length is greater than or equal to 2, return the element at index 1.

Else, return None.

```python
def second_highest_distinct(lst):
    distinct = sorted(list(set(lst)), reverse=True)
    return distinct[1] if len(distinct) >= 2 else None

# Given Test Case
print("Given Input Output:     ", second_highest_distinct([10, 20, 40, 30, 40, 20]))

# Additional Test Case
print("Additional Input Output:", second_highest_distinct([50, 50, 50]))
```

#### Q8. Remove Duplicates Without Changing the Original Order
Pseudocode:

Initialize an empty set seen and an empty list result.

Iterate through each element in the original list:

If the element is not in seen:

Add the element to seen.

Append the element to result.

Return result.

```python
def remove_duplicates_preserve_order(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

# Given Test Case
print("Given Input Output:     ", remove_duplicates_preserve_order([4, 2, 4, 5, 2, 7, 5]))

# Additional Test Case
print("Additional Input Output:", remove_duplicates_preserve_order([10, 20, 10, 30, 20, 40]))
```

#### Q9. Find the Missing Number from 1 to N
Pseudocode:

Calculate the expected sum of numbers from 1 to N using formula: N * (N + 1) // 2.

Calculate the actual sum of elements in the given list.

Subtract the actual sum from the expected sum.

Return the resulting missing number.

```python
def find_missing_number(nums, N):
    expected_sum = N * (N + 1) // 2
    return expected_sum - sum(nums)

# Given Test Case
print("Given Input Output:     ", find_missing_number([1, 2, 4, 5, 6], 6))

# Additional Test Case
print("Additional Input Output:", find_missing_number([2, 3, 4, 5], 5))
```

#### Q10. Find Two Numbers Whose Sum Equals a Target
Pseudocode:

Initialize an empty set seen to store encountered numbers.

Iterate through each number in the list from left to right:

Calculate complement: target - number.

If complement is already in seen, return [complement, number].

Add current number to seen.

If no matching pair is found, return an empty list [].

```python
def two_sum_first_pair(nums, target):
    seen = set()
    for num in nums:
        complement = target - num
        if complement in seen:
            return [complement, num]
        seen.add(num)
    return []

# Given Test Case
print("Given Input Output:     ", two_sum_first_pair([2, 7, 11, 15], 9))

# Additional Test Case
print("Additional Input Output:", two_sum_first_pair([1, 3, 5, 8], 14))
```

#### Q11. Find the Number Closest to Zero
Pseudocode:

Initialize closest with the first element of the list.

Iterate through each subsequent number in the list:

If absolute value of current number is less than absolute value of closest, update closest to current number.

Else if absolute values are equal, choose the number with the greater positive value.

Return closest.

```python
def closest_to_zero(nums):
    if not nums:
        return None
    closest = nums[0]
    for n in nums[1:]:
        if abs(n) < abs(closest):
            closest = n
        elif abs(n) == abs(closest) and n > closest:
            closest = n
    return closest

# Given Test Case
print("Given Input Output:     ", closest_to_zero([-8, -3, 2, 5, -2]))

# Additional Test Case
print("Additional Input Output:", closest_to_zero([-4, 4, -1, 1]))
```

#### Q12. Find Common Elements Between Two Lists
Pseudocode:

Convert list2 into a set set2 for fast lookup.

Initialize an empty set seen and an empty list result.

Iterate through each item in list1:

If item is present in set2 and not in seen:

Add item to seen.

Append item to result.

Return result.

```python
def common_elements(list1, list2):
    set2 = set(list2)
    seen = set()
    result = []
    for item in list1:
        if item in set2 and item not in seen:
            seen.add(item)
            result.append(item)
    return result

# Given Test Case
print("Given Input Output:     ", common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Additional Test Case
print("Additional Input Output:", common_elements([10, 20, 30, 40], [30, 50, 10, 60]))
```

#### Q13. Rotate a List to the Right by K Positions
Pseudocode:

Check if the list is empty; if empty, return [].

Compute actual rotations using modulo: k = k % len(nums).

Slice the last k elements and place them before the remaining first elements: nums[-k:] + nums[:-k].

Return the rotated list.

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k != 0 else nums[:]

# Given Test Case
print("Given Input Output:     ", rotate_right([1, 2, 3, 4, 5], 2))

# Additional Test Case
print("Additional Input Output:", rotate_right([10, 20, 30, 40, 50], 3))
```

#### Q14. Find the Maximum Sum of Two Consecutive Elements
Pseudocode:

Initialize max_sum with the sum of the first two elements (nums[0] + nums[1]).

Iterate through the list from index 1 to second-to-last element:

Calculate sum of current element and next element: current_sum = nums[i] + nums[i+1].

If current_sum is greater than max_sum, update max_sum.

Return max_sum.

```python
def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]
        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum

# Given Test Case
print("Given Input Output:     ", max_consecutive_sum([4, 7, 2, 9, 5]))

# Additional Test Case
print("Additional Input Output:", max_consecutive_sum([1, -2, 5, 8, 3]))
```

#### Q15. Find All Pairs of Numbers Whose Sum Equals a Target
Pseudocode:

Initialize empty sets seen_elements and pairs.

Iterate through each number in the list:

Compute complement: target - number.

If complement is in seen_elements:

Create a sorted tuple pair: tuple(sorted([complement, number])).

Add this pair to pairs set.

Add current number to seen_elements.

Sort all collected unique pairs by their first element.

Return the list of pairs as a list of lists.

```python
def all_pairs_with_target_sum(nums, target):
    seen_elements = set()
    pairs = set()
    for num in nums:
        complement = target - num
        if complement in seen_elements:
            pair = tuple(sorted([complement, num]))
            pairs.add(pair)
        seen_elements.add(num)
    sorted_pairs = sorted([list(p) for p in pairs], key=lambda x: x[0])
    return sorted_pairs

# Given Test Case
print("Given Input Output:     ", all_pairs_with_target_sum([2, 4, 3, 5, 7, 8, 1], 9))

# Additional Test Case
print("Additional Input Output:", all_pairs_with_target_sum([1, 5, 3, 3, 2, 4], 6))
```
