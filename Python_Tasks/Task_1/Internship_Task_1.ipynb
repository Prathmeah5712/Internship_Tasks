```python
import numpy as np

product_id = np.array([101, 102, 103, 104, 105,
                       106, 107, 108, 109, 110])

price = np.array([25000, 45000, 15000, 60000, 35000,
                  20000, 55000, 30000, 40000, 50000])

quantity = np.array([4, 2, 6, 1, 3,
                     5, 2, 7, 3, 2])

discount = np.array([5, 10, 0, 15, 5,
                     10, 5, 0, 10, 15])
```

```python
# Calculate the total sales for each product using price × quantity.
total_sales = price*quantity
print(total_sales)
```

```python
# Calculate the average total sales.
print(total_sales.mean())
```

```python
# Find the maximum and minimum total sales.
print(total_sales.max())
```

```python
print(total_sales.min())
```

```python
# Find the Product ID with the highest total sales.
np.max(product_id)
```

```python
# Find all products whose total sales are greater than the average sales.
total_sales = price * quantity
avg_sales = total_sales.mean()
products_above_avg = total_sales[total_sales > avg_sales]
print(products_above_avg)
```

```python
# Find all products where the quantity sold is greater than 4.
products_qty_gt_4 = quantity[quantity > 4]
print(products_qty_gt_4)
```

```python
# Calculate the price after applying the given discount percentage.
discounted_price = price - (price * (discount/100))
discounted_price
```

```python
# Classify each product as "High Sales" if total sales are greater than ₹1,00,000, otherwise "Low Sales" using np.where()
sales_cat = np.where(total_sales > 100000, 'High sales', 'Lowe sales')
sales_cat
```

```python
# Sort the total sales in ascending order.
sort_values = np.sort(total_sales)
sort_values
```

```python
# Find the indexes that would sort the total sales using np.argsort().
arg_sort = np.argsort(total_sales)
arg_sort
```

```python
# Display the Product IDs according to the sorted sales order.
```

```python
sales_ord = np.argsort(total_sales)
```

```python
product_id[sales_ord]
```

```python
# Check whether any product has sold more than 6 units using np.any().
sold_morethan_6 = np.any(quantity > 6)
print(sold_morethan_6)
```

```python
# Calculate the mean, median, variance, and standard deviation of the total sales.
sales_mean = np.mean(total_sales)
sales_median = np.median(total_sales)
sales_var = np.var(total_sales)
sales_std = np.std(total_sales)

print("Mean:", sales_mean)
print("Median:", sales_median)
print("Variance:", sales_var)
print("Standard Deviation:", sales_std)
```

```python
# Calculate the correlation between price and quantity sold using np.corrcoef().
corr_coef = np.corrcoef(price, quantity)
corr_coef
```
